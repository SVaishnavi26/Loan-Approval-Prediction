```python
import os
import pandas as pd

print(f"Pandas version: {pd.__version__}")
```

### 📁 Loading the Dataset
We load `loan_approval_dataset.csv` from the `../data/` directory.

> **Note for Beginners:** In raw datasets, column names and text entries often have accidental leading or trailing spaces. We remove these whitespaces using `.str.strip()` so columns can be indexed cleanly without issues later.


```python
# Load CSV file
data_path = os.path.join('..', 'data', 'loan_approval_dataset.csv')
df = pd.read_csv(data_path)

# Clean column headers (strip leading/trailing whitespaces)
df.columns = df.columns.str.strip()

# Clean categorical string columns (strip leading/trailing whitespaces)
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].str.strip()

print("Dataset loaded and cleaned successfully!")
```

### 1 & 2. Number of Rows and Columns

```python
num_rows, num_columns = df.shape
print(f"1. Total Number of Rows    : {num_rows:,}")
print(f"2. Total Number of Columns : {num_columns}")
```

### 3. All Column Names

```python
print("All Column Names:")
for i, col in enumerate(df.columns, 1):
    print(f"  {i:2d}. {col}")
```

### 4. First 5 Rows
Let's preview the first 5 records of the dataset.

```python
df.head()
```

### 5. Data Types
Understanding the data type (`int64`, `float64`, or `object` for strings) tells us which features are numerical and which are categorical.

```python
print("Data types of each column:")
print(df.dtypes)
```

### 6. Missing Values
Checking if any fields have null or missing (`NaN`) values.

```python
missing_counts = df.isnull().sum()
print("Missing values per column:")
print(missing_counts)
print(f"\nTotal missing values in entire dataset: {missing_counts.sum()}")
```

### 7. Duplicate Rows
Checking if any identical applicant records appear multiple times.

```python
duplicate_count = df.duplicated().sum()
print(f"Total duplicate rows: {duplicate_count}")
```

### 8. Unique Values of Categorical Columns
Checking the distinct options/categories present in each text column.

```python
cat_cols = df.select_dtypes(include=['object']).columns.tolist()
print("Categorical columns and their unique values:\n")
for col in cat_cols:
    uniques = df[col].unique().tolist()
    print(f"- Column: '{col}'")
    print(f"  Unique Values ({len(uniques)}): {uniques}\n")
```

### 9 & 10. Target Column and Value Distribution
The target column is **`loan_status`**, which indicates whether a loan was approved or rejected.

```python
target_col = 'loan_status'
print(f"Target Column: '{target_col}'\n")

counts = df[target_col].value_counts()
proportions = df[target_col].value_counts(normalize=True) * 100

summary_df = pd.DataFrame({
    'Count': counts,
    'Percentage (%)': proportions.round(2)
})
print("Target Value Distribution:")
print(summary_df)
```

## 📖 Column Explanations (Simple Language for Beginners)

Here is what each column in this dataset means:

| Column Name | Data Type | Description |
|:---|:---|:---|
| **`loan_id`** | Numerical (int) | Unique identification number assigned to each loan applicant (not a predictive feature). |
| **`no_of_dependents`** | Numerical (int) | Number of family members financially dependent on the applicant. |
| **`education`** | Categorical (string) | Educational qualification of the applicant (`Graduate` or `Not Graduate`). |
| **`self_employed`** | Categorical (string) | Employment type (`Yes` if running own business/freelance, `No` if salaried/employed). |
| **`income_annum`** | Numerical (int) | Annual total earnings of the applicant (in Indian Rupees ₹). |
| **`loan_amount`** | Numerical (int) | Total amount of loan requested by the applicant (in ₹). |
| **`loan_term`** | Numerical (int) | Loan repayment period in years (e.g., 2, 4, 10, 20 years). |
| **`cibil_score`** | Numerical (int) | Credit score measuring creditworthiness (ranges from 300 to 900). Higher scores mean better repayment history. |
| **`residential_assets_value`** | Numerical (int) | Total market value of residential homes/properties owned by the applicant. |
| **`commercial_assets_value`** | Numerical (int) | Total market value of commercial properties/shops owned by the applicant. |
| **`luxury_assets_value`** | Numerical (int) | Total value of luxury possessions owned (cars, jewelry, paintings, etc.). |
| **`bank_asset_value`** | Numerical (int) | Liquid cash and balance held across the applicant's bank accounts. |
| **`loan_status`** *(Target)* | Categorical (string) | The outcome to predict: **`Approved`** or **`Rejected`**. |
